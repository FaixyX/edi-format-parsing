```python
class ARR_Waterfall:
    def __init__(self, location: str, freq: str, refSheet: str, ltm_month: int = 12, events: list = None):
        # assign provided values (avoid accidental assignment to types)
        self.location = location
        self.freq = freq
        self.ltm_month = ltm_month
        self.refSheet = refSheet
        # avoid mutable default list
        self.events = events if events is not None else ['New', 'Upsell', 'Downsell', 'Churn']
    
    def collapse_to_cols(self, addrDf: pd.DataFrame):
        columns = []
        # Skip the first column (index) if it exists
        for col in addrDf.columns:
            # Get the first and last cell addresses in the column
            first_cell = addrDf.iloc[0][col]
            last_cell = addrDf.iloc[-1][col]
            # Create the address range (no trailing parenthesis)
            address = f"{first_cell}:{last_cell}"
            columns.append(address)

        return columns
    
    def build_Y_labels(self):
        events = self.events
        labels = ["Beginning ARR"]
        for event in events:
            labels.append(event)
        labels.append("Ending ARR")
        return labels
    
    def build_event_formulas(self, freqDf: pd.DataFrame):
        columns = self.collapse_to_cols(freqDf)
        events = self.events

        waterfall = []
        colSums = ["SUM("+col+")" for col in columns]
        waterfall.append(colSums[:-1])
        for event in events:
            eventRow = []
            for i, col in enumerate(columns[1:]):
                formula = self.get_wf_event_formula(col, columns[i-1], event)
                eventRow.append(formula)
            waterfall.append(eventRow)
        waterfall.append(colSums[1:])

        return waterfall


    def get_wf_event_formula(self, currCol: str, prevCol: str, event: str):
        refSheet = self.refSheet

        #make sure the sheet name is passed in formula
        currCol = "'" + refSheet + "'!" + currCol
        prevCol = "'" + refSheet + "'!" + prevCol
        
        if event == "New":
            return "=SUM((" + currCol + "-" + prevCol + ")*(" + currCol + ">" + prevCol + ")*(" + prevCol + "=0))"
        elif event == "Upsell":
            return "=SUM((" + currCol + "-" + prevCol + ")*(" + currCol + ">" + prevCol + ")*(" + prevCol + "<>0))"
        elif event == "Downsell":
            return "=SUM((" + currCol + "-" + prevCol + ")*(" + currCol + "<" + prevCol + ")*(" + currCol + "<>0))"
        elif event == "Churn":
            return "=SUM((" + currCol + "-" + prevCol + ")*(" + currCol + "<" + prevCol + ")*(" + currCol + "=0))"
        else:
            raise ValueError("Event given must equal 'New', 'Upsell', 'Downsell' or 'Churn'")
    
    def apply_freq_to_data(self, addrDf: pd.DataFrame):
        
        if self.freq == "Quarterly":
            freqDf = addrDf[addrDf.columns[addrDf.columns.is_quarter_end]]
        elif self.freq == "Annual":
            freqDf = addrDf[addrDf.columns[addrDf.columns.is_year_end]]
        elif self.freq == "LTM":
            freqDf = addrDf[addrDf.columns[addrDf.columns.month == self.ltm_month]]
        else:
            freqDf = addrDf
        return freqDf
```

```python
def num2col(x):
    """Takes a numeric input and converts it into an Excel-style column label by repeatedly dividing the 
    input by 26 and mapping the remainders to letters from 
    'A' to 'Z'. It then returns the resulting label"""
    s = list()
    while x:
        x-=1
        s.append(string.ascii_uppercase[x % 26])
        x//=26
    return "".join(reversed(s))

def col2num(col):
    """Takes an Excel column input and outputs its number"""
    num = 0
    for c in col:
        if c in string.ascii_letters:
            num = num * 26 + (ord(c.upper()) - ord('A')) + 1
    return num

def getR1C1 (cellAddr):
    """"Takes cell addresses and returns them as their numerical row and column index as an integer"""
    rowStr = ""
    colStr = ""
    for char in cellAddr:
        if char.isalpha():
            colStr += char
        elif char.isnumeric():
            rowStr += char
    return int(float(rowStr)), col2num(colStr)


def arrEventFormula (currCell, prevCell, event):
    if event == "new":
        return "=IF(AND(" + prevCell + "=0," + currCell +">" + prevCell + ")," + currCell + "-" + prevCell + ",0)"
    elif event == "upsell":
        return "=IF(AND(" + prevCell + "<>0," + currCell +">" + prevCell + ")," + currCell + "-" + prevCell + ",0)"
    elif event == "downsell":
        return "=IF(AND(" + currCell + "<>0," + currCell +"<" + prevCell + ")," + currCell + "-" + prevCell + ",0)"
    else:
        return "=IF(AND(" + currCell + "=0," + currCell +"<" + prevCell + ")," + currCell + "-" + prevCell + ",0)"
    
def createFormulaDf (event, addrDf):
    #takes formula function for an ARR event and an adress Df and returns a df of formula for the subsequent ARR event 
    colsL = len(addrDf.columns)-1 #one less because we exclude the first period bc it has no prev period to look back at
    rowsL = len(addrDf.index)
    form2dList = make2dList(rowsL, colsL)
    for colInd in range(colsL):
        for row in addrDf.itertuples(): 
            rowInd = addrDf.index.to_list().index(row[0]) #need to get its true position in case index != n
            form2dList[rowInd][colInd] = arrEventFormula(row[colInd +2],row[colInd+1], event)
    return pd.DataFrame(form2dList)

def make2dList (rowDim, colDim):
    return [[0]* colDim for i in range(rowDim)]


def range_to_dataframe(range_str: str):
    """
    Given an Excel range string (e.g., 'B2:CT135'),
    return a DataFrame of all cell addresses in that range.
    Uses only xlwings (no Excel connection).
    """

    # Parse the range
    start_row, start_col = getR1C1(range_str.split(':')[0])
    print(start_col, start_row)
    end_row, end_col = getR1C1(range_str.split(':')[1])
    print(end_col, end_row)
   

    # Generate all cell addresses
    data = []
    for row in range(start_row, end_row + 1):
        row_cells = [f"{num2col(col)}{row}" for col in range(start_col, end_col + 1)]
        data.append(row_cells)

    return pd.DataFrame(data)
```
